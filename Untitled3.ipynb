```python
import os
import glob

# 1. Update this to the folder where you extracted this specific Roboflow dataset
DATASET_DIR = '/content/Minecraft-Blocks-1'

# 2. Maps the 5 classes from this specific Roboflow universe structure to your master 7-class system
CLASS_REMAP = {
    0: 0,   # coal    -> coal ore    (Index 0)
    1: 2,   # gold    -> gold ore    (Index 2)
    2: 1,   # diamond -> diamond ore (Index 1)
    3: 6,   # emerald -> emerald ore (Index 6)
    4: 5    # iron    -> iron ore    (Index 5)
}

def align_new_roboflow_dataset(base_dir, remap_dict):
    splits = ['train', 'test', 'valid']
    print(f"[*] Aligning new Roboflow dataset structure under: {base_dir}")

    for split in splits:
        folder_path = os.path.join(base_dir, split, 'labels')
        txt_files = glob.glob(os.path.join(folder_path, "*.txt"))

        if not txt_files:
            print(f"[-] No label files found in split '{split}' ({folder_path}). Skipping.")
            continue

        modified_count = 0

        for file_path in txt_files:
            modified_lines = []
            has_changed = False

            with open(file_path, 'r') as f:
                lines = f.readlines()

            for line in lines:
                parts = line.strip().split()
                if not parts or len(parts) != 5:
                    modified_lines.append(line)
                    continue

                old_class_id = int(parts[0])

                if old_class_id in remap_dict:
                    new_class_id = remap_dict[old_class_id]
                    if old_class_id != new_class_id:
                        has_changed = True

                    # Update only index 0, keeping parts[1:] coordinates untouched
                    parts[0] = str(new_class_id)
                    modified_lines.append(" ".join(parts) + "\n")
                else:
                    modified_lines.append(line)

            if has_changed:
                with open(file_path, 'w') as f:
                    f.writelines(modified_lines)
                modified_count += 1

        print(f"[+] '{split}' update finished. Remapped classes across {modified_count} files.")

    print("\n[+] Dataset remapping completed successfully.")

# Execute the mapping transformation
align_new_roboflow_dataset(DATASET_DIR, CLASS_REMAP)
```
