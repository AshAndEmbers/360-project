```python
import os
import glob

BASE_DATASET_DIR = '/content/minecraft-2'

CLASS_REMAP = {
    0: 0,   # coal ore -> coal ore
    1: 1,   # copper ore -> copper ore
    2: 1,   # deepslate copper ore -> copper ore
    3: 2,   # deepslate diamond ore -> diamond ore
    4: 3,   # deepslate gold ore -> gold ore
    5: 4,   # deepslate lapis ore -> lapis ore
    6: 5,   # deepslate redstone ore -> redstone ore
    7: 6,   # deepslateiron ore -> iron ore
    8: 2,   # diamond ore -> diamond ore
    9: 7,   # emerald ore -> emerald ore
    10: 3,  # gold ore -> gold ore
    11: 6,  # iron ore -> iron ore
    12: 4,  # lapis ore -> lapis ore
    13: 5   # redstone ore -> redstone ore
}

def remap_entire_yolo_dataset(base_dir, remap_dict):

    splits = ['train', 'test', 'valid']

    print(f"[*] Starting dataset-wide remapping under: {base_dir}")

    for split in splits:
        folder_path = os.path.join(base_dir, split, 'labels')

        txt_files = glob.glob(os.path.join(folder_path, "*.txt"))

        if not txt_files:
            print(f"[-] No .txt label files found in split '{split}' ({folder_path}). Skipping.")
            continue

        print(f"\n[*] Processing '{split}' split: Found {len(txt_files)} files. Starting conversion...")
        modified_count = 0

        for file_path in txt_files:
            modified_lines = []
            has_changed = False

            with open(file_path, 'r') as f:
                lines = f.readlines()

            for line in lines:
                parts = line.strip().split()
                if not parts or len(parts) != 5:
                    modified_lines.append(line)
                    continue

                old_class_id = int(parts[0])

                if old_class_id in remap_dict:
                    new_class_id = remap_dict[old_class_id]
                    if old_class_id != new_class_id:
                        has_changed = True

                    parts[0] = str(new_class_id)
                    modified_lines.append(" ".join(parts) + "\n")
                else:
                    modified_lines.append(line)

            if has_changed:
                with open(file_path, 'w') as f:
                    f.writelines(modified_lines)
                modified_count += 1

        print(f"[+] '{split}' Split Complete! Correctly updated coordinate files across {modified_count} items.")

    print("\n[+] All splits processed completely.")
remap_entire_yolo_dataset(BASE_DATASET_DIR, CLASS_REMAP)

```

```python
import os
import glob


DATASET_DIR = '/content/Minecraft-Ore-Detection-2'

CLASS_REMAP = {
    0: 2,   # Deepslate Diamond Ore -> diamond ore (Index 2)
    1: 3,   # Deepslate Gold Ore -> gold ore (Index 3)
    2: 6,   # Deepslate Iron Ore -> iron ore (Index 6)
    3: 5,   # Deepslate RedstoneOre -> redstone ore (Index 5)
    4: 6    # Iron Ore -> iron ore (Index 6)
}

def remap_dataset_splits(base_dir, remap_dict):
    """
    Loops through train, valid, and test splits to safely remap class labels
    while leaving bounding box spatial coordinates perfectly intact.
    """
    splits = ['train', 'valid', 'test']

    for split in splits:
        folder_path = os.path.join(base_dir, split, 'labels')
        txt_files = glob.glob(os.path.join(folder_path, "*.txt"))

        if not txt_files:
            print(f"[-] No label files found in split: {split} ({folder_path}). Skipping.")
            continue

        print(f"[*] Processing '{split}' split: Found {len(txt_files)} files...")
        modified_count = 0

        for file_path in txt_files:
            modified_lines = []
            has_changed = False

            with open(file_path, 'r') as f:
                lines = f.readlines()

            for line in lines:
                parts = line.strip().split()
                if not parts or len(parts) != 5:
                    modified_lines.append(line)
                    continue

                old_class_id = int(parts[0])

                if old_class_id in remap_dict:
                    new_class_id = remap_dict[old_class_id]
                    if old_class_id != new_class_id:
                        has_changed = True

                    parts[0] = str(new_class_id)
                    modified_lines.append(" ".join(parts) + "\n")
                else:
                    modified_lines.append(line)

            if has_changed:
                with open(file_path, 'w') as f:
                    f.writelines(modified_lines)
                modified_count += 1

        print(f"[+] '{split}' update finished. Updated mapping inside {modified_count} files.")


remap_dataset_splits(DATASET_DIR, CLASS_REMAP)
```
